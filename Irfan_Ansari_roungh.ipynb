```python
battle=battle.assign(
    event_date = lambda x : pd.to_datetime(x['event_time']).dt.date,
    event_timex = lambda x : pd.to_datetime(x['event_time']).dt.time
)
```

```python
battle
```

```python
battle['seconds_since_midnight'] = (
    pd.to_timedelta(
        battle['event_timex'].astype(str)
    ).dt.total_seconds()
)

# Create 12 bins of 2 hours each
battle['time_span'] = (
    (battle['seconds_since_midnight'] // 7200).astype(int)
)

# Create readable labels
battle['time_span'] = battle['time_span'].apply(
    lambda x: f'{x*2:02d}:00–{(x+1)*2:02d}:00'
)
```

```python
import numpy as np
battle['day_type'] = np.where(
    pd.to_datetime(battle['event_time']).dt.dayofweek >= 5,
    'Weekend',
    'Weekday'
)
```
